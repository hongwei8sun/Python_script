```python
import pandas as pd
import xarray as xr
```

# For WMO_1st

```python

cols = ["lon","lat","z_mean","p_mean","T_mean","H2O_mean","O3_mean",
        "z_std","p_std","T_std","H2O_std","O3_std","npoints","freq"]

# Map months to seasons
month_to_season = {12:"DJF", 1:"DJF", 2:"DJF",
                   3:"MAM", 4:"MAM", 5:"MAM",
                   6:"JJA", 7:"JJA", 8:"JJA",
                   9:"SON", 10:"SON", 11:"SON"}

# Prepare seasonal containers
season_dict = {"DJF":[], "MAM":[], "JJA":[], "SON":[]}

# Load one month at a time (memory-efficient)
for m in range(1,13):
    file = f"./Raw_Data/wmo_1st/era5_wmo_1st_{m:02d}.tab"
    df = pd.read_csv(file, comment="#", sep="\s+", names=cols)
    ds = df.set_index(["lat","lon"]).to_xarray()
    ds = ds.assign_coords(month=m)
    season = month_to_season[m]
    season_dict[season].append(ds)

# Compute seasonal mean and save
for season, ds_list in season_dict.items():
    # Stack months and take mean
    ds_season = xr.concat(ds_list, dim="month").mean("month")
    
    # Convert back to DataFrame
    df_season = ds_season.to_dataframe().reset_index()
    
    # Reorder columns to match original
    df_season = df_season[cols]
    
    # Output to tab-delimited file
    out_file = f"./era5_wmo_1st_{season}.tab"
    df_season.to_csv(out_file, sep=" ", index=False, float_format="%.6f")
    
    print(f"Saved seasonal file: {out_file}")
```
